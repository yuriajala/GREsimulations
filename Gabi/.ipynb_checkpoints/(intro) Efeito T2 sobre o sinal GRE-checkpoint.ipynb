## O EFEITO DE T2* SOBRE O O SINAL GRE DO FÍGADO
Neste caderno serão estudados os efeitos causados pelo decaimento aparente T2* sobre o sinal gradiente eco (GRE) em módulo na aquisição in-phase-out-of-phase (IP-OP).

### Fundamentação Teórica
O sinal GRE é dado como uma soma de sinais complexos advindas de cada parte molecular constituinte do tecido estudado. Matematicamente, o sinal é expresso pela [equação abaixo] [1] :
[1]: https://www.ncbi.nlm.nih.gov/pubmed/18093781?dopt=Abstract "Bydder et al"

$$ S(TR,\alpha,TE) =  \left |  \;  k\sum_{n} \;\;  \rho_{n} \;\; \frac{sin[1-exp(-TR/T1_{n})]}{1-cos\alpha\cdot exp(-TR/T1_{n})} \;\; exp(-TE/T2_{n}^{*} \;\; exp(2\pi if_{n}TE)  \;  \right |$$

Onde:
* $\rho_{n}$ é a densidade de protons da parte molecular $n$ correspondente;
* $TR$ é o parâmetro tempo de repetição, referente à aquisição;
* $T1$ é propriedade de relaxação longitudinal dos spins da parte molecular $n$ do tecido;
* $TE$ é o tempo ao eco, referente à aquisição;
* $T2^{*}$ é propriedade de relaxação transversal dos spins da parte molecular $n$ do tecido.
