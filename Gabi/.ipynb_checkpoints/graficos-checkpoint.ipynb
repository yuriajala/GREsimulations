```python
import numpy as np
import skfuzzy as fuzz
import matplotlib.pyplot as plt

# Generate universe variables
#   * Quality and service on subjective ranges [0, 10]
#   * Tip has a range of [0, 25] in units of percentage points
x_OP = np.linspace(0, 1, 1500)
x_IP = np.linspace(0, 1, 1500)
x_out  = np.linspace(0, 1, 1500)

# Generate fuzzy membership functions
OP_boundary = fuzz.trapmf(x_OP, [0, 0, 0.15*0.5, 0.6*0.5])
OP_liver = fuzz.trimf(x_OP, [0.5*0.5, 0.5, 1.4*0.5])
OP_vessel = fuzz.trapmf(x_OP, [1.3*0.5, 1.6*0.5, 1, 1])

IP_vessel = fuzz.trapmf(x_IP, [0, 0, 0.15*0.5, 0.9*0.5])
IP_liver = fuzz.trimf(x_IP, [0.87*0.5, 0.5, 1.3*0.5])
IP_boundary = fuzz.trapmf(x_IP, [1.2*0.5, 1.6*0.5, 1, 1])

out_boundary = fuzz.trapmf(x_out, [0, 0, 0.05, 0.1])
out_vessel = fuzz.trimf(x_out, [0.45, 0.5, 0.55])
out_liver = fuzz.trapmf(x_out, [0.9, 0.95, 1, 1])



#    pixelIN_sm = fuzz.trapmf(x_pixelIN, [0, 0, 0.15*0.5, 0.6*0.5])
#    pixelIN_md = fuzz.trimf(x_pixelIN, [0.5*0.5, 0.5, 1.4*0.5])
#    pixelIN_la = fuzz.trapmf(x_pixelIN, [1.3*0.5, 1.6*0.5, 1, 1])
    
#   pixelIN_sm = fuzz.trapmf(x_pixelIN, [0, 0, 0.15*0.5, 0.9*0.5])
#   pixelIN_md = fuzz.trimf(x_pixelIN, [0.87*0.5, 0.5, 1.3*0.5])
#   pixelIN_la = fuzz.trapmf(x_pixelIN, [1.2*0.5, 1.6*0.5, 1, 1])
    
    
    
#    pixelOUT_sm = fuzz.trapmf(x_pixelOUT, [0, 0, 0.05, 0.1])
#    pixelOUT_md = fuzz.trimf(x_pixelOUT, [0.45, 0.5, 0.55])
#    pixelOUT_la = fuzz.trapmf(x_pixelOUT, [0.9, 0.95, 1, 1])




# Visualize these universes and membership functions
fig, (ax0, ax1, ax2) = plt.subplots(nrows=3, figsize=(8, 9))

ax0.plot(x_OP, OP_boundary, 'b', linewidth=1.5, label='Boundary')
ax0.plot(x_OP, OP_liver, 'g', linewidth=1.5, label='Liver')
ax0.plot(x_OP, OP_vessel, 'r', linewidth=1.5, label='Vessel')
ax0.set_title('Opposed-phase input membership functions')
ax0.legend()

ax1.plot(x_IP, IP_vessel, 'b', linewidth=1.5, label='Vessel')
ax1.plot(x_IP, IP_liver, 'g', linewidth=1.5, label='Liver')
ax1.plot(x_IP, IP_boundary, 'r', linewidth=1.5, label='Boundary')
ax1.set_title('In-phase input membership functions')
ax1.legend()

ax2.plot(x_out, out_boundary, 'b', linewidth=1.5, label='Boundary')
ax2.plot(x_out, out_vessel, 'g', linewidth=1.5, label='Vessel')
ax2.plot(x_out, out_liver, 'r', linewidth=1.5, label='Liver')
ax2.set_title('Output membership functions')
ax2.set_xlabel('Voxel intensity (arbitrary units)')
ax2.set_ylabel('Degree of membership')
ax2.legend()

# Turn off top/right axes
for ax in (ax0, ax1, ax2):
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.get_xaxis().tick_bottom()
    ax.get_yaxis().tick_left()

plt.tight_layout()
plt.show()
```
